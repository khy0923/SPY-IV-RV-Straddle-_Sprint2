```python
import pandas as pd
import matplotlib.pyplot as plt

mfiv = pd.read_parquet("../data/processed/mfiv_daily.parquet")
vix = pd.read_csv("../data/raw/VIXCLS.csv")
vix.columns = ["date", "vix"]
vix["date"] = pd.to_datetime(vix["date"])
vix["vix"] = pd.to_numeric(vix["vix"], errors="coerce")
vix = vix.set_index("date")["vix"].dropna()

d = pd.concat([mfiv["mfiv_vol"] * 100, vix], axis=1, join="inner").dropna()
d.columns = ["mfiv", "vix"]

print("비교 일수:", len(d))
print("상관계수(수준):", round(d.corr().iloc[0, 1], 4))
print("상관계수(일별 변화):", round(d.diff().corr().iloc[0, 1], 4))
print("평균 차이(MFIV - VIX):", round((d.mfiv - d.vix).mean(), 2), "%p")

missing = vix.loc["2010":"2023"].index.difference(mfiv.dropna().index)
print("VIX는 있는데 MFIV가 없는 날:", len(missing))

d.plot(figsize=(12, 4), title="SPY MFIV vs VIX")
plt.show()
```
